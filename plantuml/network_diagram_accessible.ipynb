# Accessible PlantUML Network and Graph Diagrams

This notebook demonstrates how to create an accessibility-aware network diagram in PlantUML from Python.

## What this diagram is used for

Network and graph diagrams represent nodes and relationships and are commonly used in cybersecurity, distributed systems, networking, and graph algorithms.

## How to make it more accessible

Use meaningful node names, explicitly label relationships, avoid unnecessary visual complexity, and provide an adjacency-style textual explanation that states which nodes are connected and what each connection represents.

## Setup

These notebooks use **Python to write PlantUML source code and call PlantUML locally** to render the diagram.

PlantUML must therefore be available on your system. A local PlantUML installation requires Java, and some diagram types may also require Graphviz.

You can use PlantUML in either of these ways:

1. **Install the PlantUML command-line tool** using the package manager available on your operating system.
2. **Download `plantuml.jar`** from the PlantUML website and run it with Java.

After installation, verify that PlantUML is available by running:

```bash
plantuml -version
```

If you use a downloaded `plantuml.jar` instead of a `plantuml` command, set the `PLANTUML_JAR` environment variable to the path of the JAR file before running the notebook.

The Python helper below supports both approaches.

````python
from pathlib import Path
import os
import shutil
import subprocess

from IPython.display import Image, display


def render_plantuml(source, filename):
    """Save PlantUML source, render it as PNG, and display it."""
    source_path = Path(f"{filename}.puml")
    source_path.write_text(source, encoding="utf-8")

    plantuml_executable = shutil.which("plantuml")
    plantuml_jar = os.environ.get("PLANTUML_JAR")

    if plantuml_executable:
        command = [
            plantuml_executable,
            "-charset", "UTF-8",
            "-tpng",
            str(source_path),
        ]
    elif plantuml_jar:
        command = [
            "java",
            "-jar",
            plantuml_jar,
            "-charset", "UTF-8",
            "-tpng",
            str(source_path),
        ]
    else:
        raise RuntimeError(
            "PlantUML was not found. Install the 'plantuml' command "
            "or set PLANTUML_JAR to the path of plantuml.jar."
        )

    result = subprocess.run(command, capture_output=True, text=True)

    if result.returncode != 0:
        raise RuntimeError("PlantUML rendering failed:\n" + result.stderr)

    image_path = source_path.with_suffix(".png")
    if not image_path.exists():
        raise FileNotFoundError(f"Rendered image was not created: {image_path}")

    display(Image(filename=str(image_path)))
    return image_path
````

## Accessible example

The example below creates an accessibility-aware network diagram also presented in the Guidelines document.

````python
network_source = """@startuml
skinparam dpi 300
node Client
node WebServer
node Database

Client --> WebServer : HTTP request
WebServer --> Database : SQL query
@enduml
"""

render_plantuml(network_source, "accessible_network_diagram")
````

## Accessible description

Network diagram with three nodes: Client, WebServer, and Database. Client connects to WebServer through an HTTP request. WebServer connects to Database through an SQL query. The structure represents a simple three-tier client-server architecture.

## Reusable accessible template

Replace the placeholder node and edge names with meaningful system components and explicit relationship labels.

````python
network_template = """@startuml
node SourceNode
node IntermediateNode
node DestinationNode

SourceNode --> IntermediateNode : Describe first connection
IntermediateNode --> DestinationNode : Describe second connection
@enduml
"""

render_plantuml(network_template, "network_diagram_template")
````

## Accessible description template

After replacing the placeholder source with your own network diagram, add a concise textual description that:

- identifies the diagram type and purpose,
- names the important entities, classes, actions, states, or nodes,
- explains the relationships or interaction order,
- describes decision or transition logic where relevant,
- and states the key structural takeaway.

Do not rely on spatial wording alone. The description should communicate the important relationships encoded by the diagram.
