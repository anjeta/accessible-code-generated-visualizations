# Generating Alternative Text for Existing Graphs with Chatlas and Gemini

This notebook demonstrates how to use **Chatlas with Gemini** as a semantic interpretation layer for graphs that already exist as image files. The aim is to generate natural-language descriptions that support blind and visually impaired users.

## Workflow

The supplied material presents Chatlas as a tool for generating accessible explanations of plots and diagrams. For an existing PNG graph, the workflow is:

1. Create a `ChatGoogle` client using Gemini.
2. Provide the path to the PNG with `content_image_file(...)`.
3. Ask for a description for a visually impaired user.
4. Convert the response to text.
5. Save the description to a `.txt` file.

## Imports and Gemini client

Replace `YOUR_API_KEY` with your Gemini API key.

```python
from chatlas import content_image_file, ChatGoogle

chat = ChatGoogle(model="gemini-2.5-flash", api_key="YOUR_API_KEY")
```

## Function for an existing PNG graph

The example first loads a saved Matplotlib figure and then describes the resulting PNG. Because the graph already exists as a PNG, the function begins directly from its file path. 

The image extension does not have to be PNG.

```python
def generate_alt_text_from_png(img_path, output_txt="alttext_report.txt"):
    response = chat.chat(
        "Describe this chart for a visually impaired user.",
        content_image_file(img_path)
    )

    alt_text = str(response)

    with open(output_txt, "w") as f:
        f.write(alt_text)

    print(f"Analysis complete. Created {output_txt}")
    return alt_text
```

## Example

Replace the placeholder path with the path to an existing graph.

```python
image_path = "path/to/your_graph.png"
output_file = "your_graph_alt_text.txt"

alt_text = generate_alt_text_from_png(image_path, output_txt=output_file)
print("\nGenerated alternative text:\n")
print(alt_text)
```

## Reusable placeholder

For another graph, change only the path and output filename.

```python
image_path = "PATH_TO_EXISTING_GRAPH.png"
output_file = "ALT_TEXT_OUTPUT.txt"

alt_text = generate_alt_text_from_png(image_path, output_txt=output_file)
```
