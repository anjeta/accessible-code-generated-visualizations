# Accessible Line Graphs in R

This notebook is a standalone teaching resource for creating accessible line plots with **ggplot2** in R.

## What this graph is used for

Line graphs are used to represent change over a continuous variable, often time or input size, as well as functional relationships, trends, and growth patterns.

## How to make it more accessible

Use meaningful variable names, descriptive axis labels, a descriptive title, readable font sizes, grid lines, and high-contrast styling. Do not rely on color alone: combine the line with point markers so that the series remains distinguishable non-visually and in grayscale.

## Accessible example

The code below showcases an accessibility-aware example of a line graph created in R.

````r
epochs <- 1:10
training_loss <- c(
  0.95, 0.82, 0.71, 0.60, 0.52,
  0.45, 0.40, 0.37, 0.35, 0.34
)

line_data <- data.frame(
  epoch = epochs,
  training_loss = training_loss
)

ggplot(line_data, aes(x = epoch, y = training_loss)) +
  geom_line(
    color = "black",
    linewidth = 1
  ) +
  geom_point(
    color = "black",
    shape = 16,
    size = 3
  ) +
  labs(
    title = "Neural Network Training Loss Across Epochs",
    x = "Training Epoch",
    y = "Loss Value"
  ) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(size = 16),
    panel.grid.minor = element_blank()
  )
````
