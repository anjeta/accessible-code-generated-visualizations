# Accessible Histograms in R

This notebook is a standalone teaching resource for creating accessible histograms with **ggplot2** in R.

## What this graph is used for

Histograms are used to represent frequency distributions, data spread, and skewness of a measured variable.

## How to make it more accessible

Explain the represented variable, choose the number of bins explicitly, add descriptive axis labels and a title, and use high-contrast bin borders. Grid lines and readable text can further support estimation and interpretation.

## Accessible example

The code below showcases an accessibility-aware example of a histogram created in R.

````r
exam_scores <- c(55, 60, 62, 65, 68, 70, 72, 75, 78, 80)

histogram_data <- data.frame(
  exam_score = exam_scores
)

ggplot(histogram_data, aes(x = exam_score)) +
  geom_histogram(
    bins = 5,
    fill = "lightgray",
    color = "black"
  ) +
  labs(
    title = "Distribution of Exam Scores",
    x = "Exam Score (%)",
    y = "Number of Students"
  ) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(size = 16),
    panel.grid.minor = element_blank()
  )
````

## Reusable accessible template

Replace the placeholder data, bin count, and text with your own values. Keep the bin count explicit and use clear labels, high-contrast borders, readable text, and an uncluttered grid.

````r
# Replace the placeholder values and text below with your own data.
values <- c(12, 15, 17, 18, 19, 22, 23, 25, 27, 30)
number_of_bins <- 5

x_axis_label <- "Measured variable (unit)"
y_axis_label <- "Frequency"
graph_title <- "Distribution of the measured variable"

template_data <- data.frame(
  value = values
)

ggplot(template_data, aes(x = value)) +
  geom_histogram(
    bins = number_of_bins,
    fill = "lightgray",
    color = "black"
  ) +
  labs(
    title = graph_title,
    x = x_axis_label,
    y = y_axis_label
  ) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(size = 16),
    panel.grid.minor = element_blank()
  )
````
