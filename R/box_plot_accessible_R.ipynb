# Accessible Box Plots in R

This notebook is a standalone teaching resource for creating accessible box plots with **ggplot2** in R.

## What this graph is used for

Box plots summarize a distribution using the median, quartiles, variability, and possible outliers.

## How to make it more accessible

Contextualize what the values represent, label measurement units, use a descriptive title and readable text, and include grid lines when they help users estimate values. A meaningful category label also makes the statistical summary easier to interpret.

## Accessible example

The code below showcases an accessibility-aware example of a box-plot created in R.

````r
algorithm_runtime <- c(10, 12, 14, 15, 18, 19, 20, 21, 35)

box_data <- data.frame(
  category = "Runtime",
  execution_time = algorithm_runtime
)

ggplot(box_data, aes(x = category, y = execution_time)) +
  geom_boxplot(
    fill = "lightgray",
    color = "black",
    width = 0.4
  ) +
  labs(
    title = "Distribution of Algorithm Execution Time",
    x = NULL,
    y = "Execution Time (ms)"
  ) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(size = 16),
    panel.grid.minor = element_blank()
  )
````
